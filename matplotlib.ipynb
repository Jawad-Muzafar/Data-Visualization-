## Plot with label and title

```python
plt.figure(figsize=(8, 5), dpi=140)
plt.plot(df['index'], df['RG Sharma'])
plt.plot(df['index'], df['V Kohli'])
plt.title('Runs scored by RG Sharma and V Kohli')
plt.xlabel('Match Index')
plt.ylabel('Runs scored')
plt.legend()
plt.show()
```

## Color , Line and Marker size 

```python
plt.figure(figsize=(8, 5), dpi=140)
plt.plot(df['index'], df['RG Sharma'],color='red', linestyle='dashed', marker='o', label='RG Sharma',)
plt.plot(df['index'], df['V Kohli'], color='blue', linestyle='solid', marker='s', label='V Kohli')
plt.title('Runs scored by RG Sharma and V Kohli')
plt.xlabel('Match Index')
plt.ylabel('Runs scored')
plt.legend()
plt.show()
```

## Limiting figures

```python
plt.figure(figsize=(8, 5), dpi=140)
price=[100, 200, 300, 400, 5000,]
year=[2015, 2016, 2017, 2018, 2019]
plt.plot(year, price)
```

```python
## we use another way to manage these type of graphs
plt.figure(figsize=(8, 5), dpi=140)
price=[100, 200, 300, 400, 5000]
year=[2015, 2016, 2017, 2018, 2019]
plt.plot(year, price, marker='o')
plt.xlim(2015, 2019)
plt.ylim(0, 1000)
plt.title('Limiting figures')
plt.xlabel('Year')
plt.ylabel('Price')
plt.show()
```

## Making grid 

```python
plt.figure(figsize=(8, 5), dpi=140)
plt.plot(df['index'], df['RG Sharma'],color='red', linestyle='dashed', marker='o', label='RG Sharma',)
plt.plot(df['index'], df['V Kohli'], color='blue', linestyle='solid', marker='s', label='V Kohli')
plt.title('Runs scored by RG Sharma and V Kohli')
plt.xlabel('Match Index')
plt.ylabel('Runs scored')
plt.grid()
plt.legend()
plt.show()
```

## **Histogram**

```python
#Lets make a histogram
#Sample data for histogram
plt.figure(figsize=(8, 5), dpi=140)
data = np.random.randn(100)
bins = np.arange(-4, 5, 1)  # Define bin edges from -4 to 4 with a step of 1
plt.hist(data, bins=bins)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

## **Pie chart**

```python
#sample data for pie chart
plt.figure(figsize=(8, 5), dpi=140)
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart')
plt.show()
```

```python
df4.head()
```

```python
##Pie chart color customization
plt.figure(figsize=(8, 5), dpi=140)
plt.pie(df_head3['batsman_runs'], labels=df_head3['batsman'], autopct='%1.1f%%', startangle=90, colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen'])
plt.title('Pie Chart')
plt.show()
```

```python
plt.figure(figsize=(8, 5), dpi=140)
plt.pie(df_head3['batsman_runs'], labels=df_head3['batsman'], autopct='%1.1f%%', startangle=90,explode=(0.1, 0.1), shadow=True, colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen'])
plt.title('Pie Chart')
plt.show()
```
