## **Importing Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sns.set_theme(style="whitegrid")
plt.style.use("default")
```

## Loading dataset

```python
df0=pd.read_csv('C:/Data sets/batter.csv')
```

```python
df=sns.load_dataset('tips')
```

```python
df.head(5)
```

## **Scatter plot**

```python
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex',style='time')
plt.title('Scatter plot of total bill vs tip')
plt.show()
```

## **Lineplot**

```python
gap=px.data.gapminder()
pak=gap[gap['country']=='Pakistan']
```

```python
plt.figure(figsize=(8,6),dpi=150)
sns.relplot(kind='line',data=pak,x='year',y='lifeExp')
plt.xlabel('year')
plt.ylabel('Lifeexp')
plt.show()
```

```python
con=gap[gap['country'].isin(['India','Germany','China'])]
```

```python
sns.relplot(kind='line',data=con,x='year',y='lifeExp',hue='country',style='continent')
plt.title('Life Expectancy over the years for India, Germany and China')
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.show()
```

## **Facet plot**

```python
sns.relplot(kind='scatter',data=df,x='total_bill',y='tip' ,col='sex')
plt.title('Scatter plot of total bill vs tip by sex')
plt.show()
```

## **Bar plot**

```python
sns.barplot(data=df,x='day',y='total_bill',hue='sex')
plt.title('Bar plot of total bill by day and sex')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

```python
df.head()
```

## **Count plot**

```python
sns.countplot(data=df,x='smoker',hue='time')
plt.title('Count of smokers by time of day')
plt.xlabel('Smoker status')
plt.ylabel('Count')
plt.show()
```

##  **Pair plot**

```python
sns.pairplot(data=df,hue='sex',vars=['total_bill','tip'])
plt.xlabel('Total bill')
plt.ylabel('Tip')
plt.show()
```

## **Catplot**

```python
sns.catplot(data=df,x='day',y='total_bill',hue='sex',kind='box')
plt.title('Box plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

## **Boxen plot**

```python
sns.boxenplot(data=df,x='day',y='total_bill',hue='sex')
plt.title('Boxen plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

## **Displot**

```python
sns.displot(data=df,x='total_bill',hue='sex',kde=True)
plt.title('Distribution of total bill by sex')
plt.xlabel('Total bill amount')
plt.ylabel('Density')
plt.show()
```

## **Jointplot**

```python
sns.jointplot(data=df,x='total_bill',y='tip',hue='smoker')
plt.xlabel('Total bill')
plt.ylabel('Tip')
plt.show()
```

## **Histplot**

```python
sns.histplot(data=df,x='total_bill',hue='sex',multiple='stack',bins=20)
plt.title('Histogram of total bill by sex')
plt.xlabel('Total bill amount')
plt.ylabel('Count')
plt.show()
```

## **Swarm plot**

```python
sns.swarmplot(data=df,x='day',y='total_bill',hue='sex')
plt.title('Swarm plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

## **Violin Plot**

```python
sns.violinplot(data=df,x='day',y='total_bill',hue='sex',split=True)
plt.title('Violin plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

## **kde plot**

```python
sns.kdeplot(data=df,x='total_bill',hue='sex',multiple='stack',fill=True)
plt.title('Kernel Density Estimate of total bill by sex')
plt.xlabel('Total bill amount')
plt.ylabel('Density')
plt.show()
```

## **Strip plot**

```python
sns.stripplot(data=df,x='day',y='total_bill',hue='sex',jitter=True)
plt.title('Strip plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```

```python
sns.pointplot(data=df,x='day',y='total_bill',hue='sex')
plt.title('Point plot of total bill by day')
plt.xlabel('Day of the week')
plt.ylabel('Total bill amount')
plt.show()
```
